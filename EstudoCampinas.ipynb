```python
from pysus import sinan , cnes
```

```python
arquivos = sinan(disease='deng', year=[2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024, 2025, 2026], as_dataframe=False)
```

```python
import duckdb

CODIGO_CAMPINAS = "350950"

# Converte os caminhos para uma lista válida no SQL do DuckDB
caminhos = [
    str(getattr(arquivo, "path", arquivo))
    for arquivo in arquivos
]

caminhos_sql = ", ".join(
    f"'{caminho.replace(chr(39), chr(39) * 2)}'"
    for caminho in caminhos
)

consulta = f"""
    SELECT *
    FROM read_parquet(
        [{caminhos_sql}],
        union_by_name = TRUE
    )
    WHERE LEFT(CAST(ID_MUNICIP AS VARCHAR), 6) = '{CODIGO_CAMPINAS}'
"""

df_sinan = duckdb.sql(consulta).df()

print("Dimensões:", df_sinan.shape)
display(df_sinan.head())
```

```python
df_sinan.info(verbose=True)
```

```python
colunas_casos = [
    "ID_UNIDADE",
    "SG_UF_NOT",
    "ID_MUNICIP",
    "NU_ANO",
    "DT_NOTIFIC",
    "SEM_NOT",
    "DT_SIN_PRI",
    "SEM_PRI",
    "ID_MN_RESI",
    "CLASSI_FIN",
    "CRITERIO",
    "EVOLUCAO",
    "SOROTIPO",
    "VINC_SUS"
]

# Mantém somente as colunas que realmente existem
colunas_disponiveis = [
    coluna
    for coluna in colunas_casos
    if coluna in df_sinan.columns
]

casos_campinas = df_sinan[
    colunas_disponiveis
].copy()

display(casos_campinas.head())
```

```python
import pandas as pd
import pysus

try:
    from pysus import cnes, list_files
except ImportError:
    # Fallback para algumas versões do PySUS
    from pysus.api._impl.databases import cnes, list_files

print("PySUS:", pysus.__version__)
```

```python
catalogo_cnes_2026 = list_files(
    dataset="cnes",
    state="SP",
    year=2026,
    group="ST"
)

if catalogo_cnes_2026.empty:
    raise RuntimeError("Nenhum arquivo CNES ST de 2026 foi encontrado.")

catalogo_cnes_2026["month"] = pd.to_numeric(
    catalogo_cnes_2026["month"],
    errors="coerce"
)

display(
    catalogo_cnes_2026
    .sort_values(["year", "month"])
    .tail(12)
)

mes_cnes_mais_recente = int(
    catalogo_cnes_2026["month"].dropna().max()
)

print("Competência mais recente:", f"2026-{mes_cnes_mais_recente:02d}")
```

```python
cnes_sp = cnes(
    state="SP",
    year=2026,
    month=mes_cnes_mais_recente,
    group="ST",
    as_dataframe=True,
    add_dv=False,
    show_progress=True
)

cnes_sp.columns = (
    cnes_sp.columns
    .str.upper()
    .str.strip()
)

print("Formato:", cnes_sp.shape)
display(cnes_sp.head())
```

```python
def normalizar_codigo(serie, tamanho):
    return (
        serie.astype("string")
        .str.strip()
        .str.replace(r"\.0$", "", regex=True)
        .str.replace(r"\D", "", regex=True)
        .replace("", pd.NA)
        .str.zfill(tamanho)
    )
```

```python
def encontrar_coluna(dataframe, candidatos):
    for coluna in candidatos:
        if coluna in dataframe.columns:
            return coluna
    return None


col_cnes = encontrar_coluna(
    cnes_sp,
    ["CNES", "CO_CNES", "COD_CNES"]
)

col_municipio = encontrar_coluna(
    cnes_sp,
    ["CODUFMUN", "CO_MUNICIPIO_GESTOR", "CO_MUNICIPIO"]
)

col_nome = encontrar_coluna(
    cnes_sp,
    [
        "FANTASIA",
        "NO_FANTASIA",
        "NOMEFANTA",
        "NOME_FANTASIA",
        "RAZAOSOC",
        "NO_RAZAO_SOCIAL"
    ]
)

col_latitude = encontrar_coluna(
    cnes_sp,
    ["LATITUDE", "NU_LATITUDE", "VL_LATITUDE"]
)

col_longitude = encontrar_coluna(
    cnes_sp,
    ["LONGITUDE", "LONGITUD", "NU_LONGITUDE", "VL_LONGITUDE"]
)

print("Código CNES:", col_cnes)
print("Município:", col_municipio)
print("Nome:", col_nome)
print("Latitude:", col_latitude)
print("Longitude:", col_longitude)

if col_cnes is None:
    raise KeyError("Não foi possível localizar a coluna de código CNES.")
```

```python
cnes_lookup = pd.DataFrame()

cnes_lookup["ID_UNIDADE"] = normalizar_codigo(
    cnes_sp[col_cnes],
    7
)

if col_municipio is not None:
    cnes_lookup["MUNICIPIO_UNIDADE"] = (
        normalizar_codigo(cnes_sp[col_municipio], 6)
        .str[:6]
    )
else:
    cnes_lookup["MUNICIPIO_UNIDADE"] = pd.NA

if col_nome is not None:
    cnes_lookup["NOME_UNIDADE"] = (
        cnes_sp[col_nome]
        .astype("string")
        .str.strip()
    )
else:
    cnes_lookup["NOME_UNIDADE"] = pd.NA

if col_latitude is not None:
    cnes_lookup["LATITUDE"] = pd.to_numeric(
        cnes_sp[col_latitude]
        .astype("string")
        .str.replace(",", ".", regex=False),
        errors="coerce"
    )
else:
    cnes_lookup["LATITUDE"] = pd.NA

if col_longitude is not None:
    cnes_lookup["LONGITUDE"] = pd.to_numeric(
        cnes_sp[col_longitude]
        .astype("string")
        .str.replace(",", ".", regex=False),
        errors="coerce"
    )
else:
    cnes_lookup["LONGITUDE"] = pd.NA

cnes_lookup = (
    cnes_lookup
    .dropna(subset=["ID_UNIDADE"])
    .drop_duplicates(subset=["ID_UNIDADE"])
    .reset_index(drop=True)
)

print("Estabelecimentos cadastrados:", len(cnes_lookup))
display(cnes_lookup.head())
```

```python
cnes_campinas = (
    cnes_lookup.loc[
        cnes_lookup["MUNICIPIO_UNIDADE"] == "350950"
    ]
    .copy()
    .reset_index(drop=True)
)

print("Unidades do CNES localizadas em Campinas:", len(cnes_campinas))

display(
    cnes_campinas[
        [
            "ID_UNIDADE",
            "NOME_UNIDADE",
            "LATITUDE",
            "LONGITUDE"
        ]
    ].head(20)
)
```

```python
import geopandas as gpd

from geobr import read_health_facilities
```

```python
cnes_campinas_geo = read_health_facilities(
    date=202601,
    code_muni=3509502,
    output="gpd",
    show_progress=True,
    cache=True,
    verbose=True
)

print("Formato:", cnes_campinas_geo.shape)
print("CRS:", cnes_campinas_geo.crs)
print("\nColunas:")
print(cnes_campinas_geo.columns.tolist())

display(cnes_campinas_geo.head())
```

```python
cnes_campinas_geo = cnes_campinas_geo.to_crs("EPSG:4326")

cnes_campinas_geo["LONGITUDE"] = (
    cnes_campinas_geo.geometry.x
)

cnes_campinas_geo["LATITUDE"] = (
    cnes_campinas_geo.geometry.y
)

display(
    cnes_campinas_geo[
        [
            "LATITUDE",
            "LONGITUDE",
            "geometry"
        ]
    ].head()
)
```

```python
colunas_relevantes = [
    coluna
    for coluna in cnes_campinas_geo.columns
    if any(
        termo in coluna.lower()
        for termo in [
            "cnes",
            "name",
            "nome",
            "fant",
            "facility",
            "estab",
            "coord",
            "vinc_sus"
        ]
    )
]

print(colunas_relevantes)
```

```python
def encontrar_coluna(dataframe, candidatos):
    mapa = {
        coluna.lower(): coluna
        for coluna in dataframe.columns
    }

    for candidato in candidatos:
        if candidato.lower() in mapa:
            return mapa[candidato.lower()]

    return None


col_cnes_geo = encontrar_coluna(
    cnes_campinas_geo,
    [
        "code_cnes",
        "cnes",
        "co_cnes",
        "codigo_cnes"
    ]
)

col_nome_geo = encontrar_coluna(
    cnes_campinas_geo,
    [
        "name_facility",
        "name_health_facility",
        "name_establishment",
        "nome_fantasia",
        "no_fantasia",
        "nome_estabelecimento",
        "fantasia",
        "vinc_sus"
    ]
)

print("Código CNES:", col_cnes_geo)
print("Nome:", col_nome_geo)
```

```python
import pandas as pd

# Normalizar o código de natureza jurídica
cnes_campinas_geo["COD_NATUREZA_JURIDICA"] = (
    cnes_campinas_geo["co_natureza_jur"]
    .astype("string")
    .str.strip()
    .str.replace(r"\.0$", "", regex=True)
    .str.replace(r"\D", "", regex=True)
    .replace("", pd.NA)
    .str.zfill(4)
)

# O primeiro dígito identifica o grande grupo jurídico
grupo_natureza = (
    cnes_campinas_geo["COD_NATUREZA_JURIDICA"]
    .str[0]
)

cnes_campinas_geo["CATEGORIA_NATUREZA"] = (
    grupo_natureza.map({
        "1": "Administração pública",
        "2": "Entidade empresarial",
        "3": "Entidade sem fins lucrativos",
        "4": "Pessoa física",
        "5": "Organização internacional ou extraterritorial"
    })
    .fillna("Não informado")
)

# Classificação simplificada solicitada
cnes_campinas_geo["TIPO_REDE"] = (
    grupo_natureza.map({
        "1": "SUS / Pública",
        "2": "Privada",
        "3": "Privada",
        "4": "Privada",
        "5": "Outra / Não pública"
    })
    .fillna("Não informado")
)
```

```python
def normalizar_cnes(serie):
    return (
        serie.astype("string")
        .str.strip()
        .str.replace(r"\.0$", "", regex=True)
        .str.replace(r"\D", "", regex=True)
        .replace("", pd.NA)
        .str.zfill(7)
    )


if col_cnes_geo is None:
    raise KeyError(
        "Não foi encontrada a coluna com o código CNES."
    )

cnes_final = pd.DataFrame({
    "ID_UNIDADE": normalizar_cnes(
        cnes_campinas_geo[col_cnes_geo]
    ),

    "NOME_UNIDADE": (
        cnes_campinas_geo[col_nome_geo].astype("string")
        if col_nome_geo is not None
        else pd.NA
    ),

    "COD_NATUREZA_JURIDICA": (
        cnes_campinas_geo["COD_NATUREZA_JURIDICA"]
    ),

    "CATEGORIA_NATUREZA": (
        cnes_campinas_geo["CATEGORIA_NATUREZA"]
    ),

    "TIPO_REDE": (
        cnes_campinas_geo["TIPO_REDE"]
    ),

    "LATITUDE": cnes_campinas_geo["LATITUDE"],

    "LONGITUDE": cnes_campinas_geo["LONGITUDE"]
})

if "coords_source" in cnes_campinas_geo.columns:
    cnes_final["FONTE_COORDENADA"] = (
        cnes_campinas_geo["coords_source"]
    )

cnes_final = (
    cnes_final
    .dropna(subset=["ID_UNIDADE"])
    .drop_duplicates(subset=["ID_UNIDADE"])
    .reset_index(drop=True)
)

display(cnes_final.head(20))
```

```python
df_casos = casos_campinas.copy()

df_casos["ID_UNIDADE"] = normalizar_cnes(
    df_casos["ID_UNIDADE"]
)

casos_com_unidade = df_casos.merge(
    cnes_final,
    on="ID_UNIDADE",
    how="left",
    validate="many_to_one",
    indicator="STATUS_CNES"
)

display(casos_com_unidade.head())
```
